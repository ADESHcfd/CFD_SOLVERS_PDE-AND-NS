# Truncation Error Analysis of FTCS Schemes

This notebook derives the **truncation errors**, determines the **order of accuracy**, and discusses the **stability characteristics** of the Forward Time Central Space (FTCS) finite difference scheme for the 1D Advection and Diffusion equations.

---

# 1. Governing Equations

## Advection Equation

$$
\frac{\partial u}{\partial t}
+
c\frac{\partial u}{\partial x}
=
0
$$

## Diffusion Equation

$$
\frac{\partial u}{\partial t}
=
\alpha
\frac{\partial^2u}{\partial x^2}
$$

---

# 2. FTCS Analysis for the Advection Equation

## FTCS Discretisation

Using a Forward Difference in Time and Central Difference in Space:

$$
\frac{u_i^{n+1}-u_i^n}{\Delta t}
+
c
\frac{u_{i+1}^{n}-u_{i-1}^{n}}
{2\Delta x}
=
0
$$

---

## Modified Equation

After Taylor series expansion:

$$
u_t+cu_x
=
-\frac{\Delta t}{2}u_{tt}
-\frac{c\Delta x^2}{6}u_{xxx}
+\cdots
$$

---

## Truncation Error

$$
TE
=
\frac{\Delta t}{2}u_{tt}
+
\frac{c\Delta x^2}{6}u_{xxx}
+\cdots
$$

### Accuracy

- First Order in Time: $O(\Delta t)$
- Second Order in Space: $O(\Delta x^2)$

---

## Stability

Von Neumann analysis shows that the amplification factor satisfies:

$$
|G|>1
$$

for some wave numbers regardless of timestep.

Therefore,

$$
\boxed{
\text{FTCS Advection is Unconditionally Unstable}
}
$$

---

# 3. FTCS Analysis for the Diffusion Equation

## FTCS Discretisation

$$
\frac{u_i^{n+1}-u_i^n}{\Delta t}
=
\alpha
\frac{
u_{i+1}^n
-
2u_i^n
+
u_{i-1}^n
}
{\Delta x^2}
$$

---

## Modified Equation

After Taylor series expansion:

$$
u_t-\alpha u_{xx}
=
-\frac{\Delta t}{2}u_{tt}
+
\frac{\alpha\Delta x^2}{12}u_{xxxx}
+\cdots
$$

---

## Truncation Error

$$
TE
=
\frac{\Delta t}{2}u_{tt}
-
\frac{\alpha\Delta x^2}{12}u_{xxxx}
+\cdots
$$

### Accuracy

- First Order in Time: $O(\Delta t)$
- Second Order in Space: $O(\Delta x^2)$

---

## Stability

Define the Fourier Number:

$$
Fo
=
\frac{\alpha \Delta t}{\Delta x^2}
$$

The FTCS diffusion scheme is stable if

$$
\boxed{
Fo \le 0.5
}
$$

or equivalently

$$
\boxed{
\Delta t
\le
\frac{\Delta x^2}{2\alpha}
}
$$

---

# 4. Summary

| Scheme | Time Accuracy | Space Accuracy | Stability |
|----------|----------|----------|----------|
| FTCS Advection | First Order | Second Order | Unconditionally Unstable |
| FTCS Diffusion | First Order | Second Order | Stable if $Fo \le 0.5$ |

---

# Key Takeaways

- Truncation error is obtained using Taylor series expansion.
- Both FTCS schemes are first-order accurate in time.
- Both FTCS schemes are second-order accurate in space.
- FTCS Advection is unconditionally unstable.
- FTCS Diffusion is conditionally stable.
- Stability is as important as accuracy when developing CFD solvers.
